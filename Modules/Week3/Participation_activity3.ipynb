```python
#3a
def say_hello_to(name):
  print(name)
say_hello_to("Amy")
#in this code, first I define the function "say_hello_to(name)" with print(name). This step defines the function. And next, say_hello_to("Amy") calls this function. The code in this function executes when and only when the function is called.
```

```python
#3b
#According to the Chapter 4 we read last week,  there is only one global scope, created when your program begins.
#A new local scope gets created whenever a program calls a function.
def cor(name):
  print(name)
cor("Amy")
cor("Jason")

```

```python
#3c
#When the function returns, the local scope gets destroyed, along with these variables.
def cor():
  ducks = 'sss'
cor()
print(ducks)
#at this time if I run print(ducks), there will be an error.
#The error happen because ducks variable exists only in the local scope when cor() is called. Once the program execution returns from cor(), that local scope gets destroyed, and there is no longer a variable named ducks
```

```python
#3d
#Errors can be handled with try and except statements. The code that could potentially have an error is put in a try clause. The program execution moves to the start of a following except clause if an error happens.
def num(number):
  try:
    return 63/number
  except ZeroDivisionError:
    # If ZeroDivisionError happened, the code in this block runs:
        print('Error: Invalid argument.')
print(num(3))
print(num(21))
print(num(0))
print(num(7))
```

```python
#4
def collatz(number):
  if number % 2 == 0:
    result = number // 2
  if number % 2 == 1:
    result = 3 * number + 1
  print(result,sep=' ')
  return result
number = int(input("Enter your number:"))

print(number,sep='')
while number != 1:
  number = collatz(number)
```
