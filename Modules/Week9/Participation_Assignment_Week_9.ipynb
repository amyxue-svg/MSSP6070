```python
import requests, os, bs4, time
```

```python
url = 'https://www.flickr.com/search/?text=pasta'
os.makedirs('images', exist_ok=True)
```

```python
res = requests.get(url)
res.raise_for_status()
soup = bs4.BeautifulSoup(res.text, 'html.parser')
```

```python
image_elems = soup.select('.photo-list-photo-view img')
print(len(image_elems))
```

```python
image_url = 'https:' + image_elems[0].get('src')
print(image_url)
image_res = requests.get(image_url)
image_res.raise_for_status()
```

```python
for image_elem in image_elems:
    image_url = 'https:' + image_elem.get('src')
    print('Downloading image:', image_url)
    image_res = requests.get(image_url)
    image_res.raise_for_status()
    image_file = open(
        os.path.join('images', os.path.basename(image_url)), 'wb'
    )
    for chunk in image_res.iter_content(100000):
        image_file.write(chunk)
    image_file.close()
    time.sleep(1)
print('Done.')
```
