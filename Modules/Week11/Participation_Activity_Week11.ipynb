```python
districts = gpd.read_file("data/california-congressional-districts.geojson")
```

```python
population = pd.read_csv("Modules/Week11/CA_Legislative_Districts (1).csv",
                          dtype={"DISTRICT": str})
```

```python
districts.boundary.plot(color="black", figsize=(10, 10), linewidth=0.7)
plt.title("California Congressional Districts")
plt.axis("off")
plt.show()
```

```python
district_map = districts.merge(
    population[["DISTRICT", "POPULATION"]],
    left_on="Name",
    right_on="DISTRICT",
    how="left",
)
print(district_map)
```

```python
district_map.plot(
    column="POPULATION",
    cmap="YlOrRd",
)
plt.title("California Congressional District Population")
plt.axis("off")
plt.show()
```
