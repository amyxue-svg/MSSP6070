```python
aapl = pd.read_csv(".git/data/Week 05 Examples/aapl.csv")
amzn = pd.read_csv(".git/data/Week 05 Examples/amzn.csv")
fb = pd.read_csv(".git/data/Week 05 Examples/fb.csv")
goog = pd.read_csv(".git/data/Week 05 Examples/goog.csv")
nflx = pd.read_csv(".git/data/Week 05 Examples/nflx.csv")
print(aapl)
print(amzn)
print(fb)
print(goog)
print(nflx)
```

c.

```python
aapl["ticker"] = "AAPL"
amzn["ticker"] = "AMZN"
fb["ticker"] = "FB"
goog["ticker"] = "GOOG"
nflx["ticker"] = "NFLX"
```

d.

```python
faanf = pd.concat([aapl,amzn,fb,goog,nflx])
print(faanf)
```

e.

```python
faanf.to_csv("faanf.csv")
pd.read_csv("faanf.csv")
```

f.

```python
faang = faanf.assign(
    date=pd.to_datetime(faanf.date),
    volume=faanf.volume.astype(int)
).sort_values(
    ['date', 'ticker']
)
print(faang[["date","volume","ticker"]])
faang.head()
```

g.

```python
faang.nlargest(7,"volume")
```

h.

```python
melt_faang = faang.melt(
    id_vars=['ticker', 'date'],
    value_vars=['open', 'high', 'low', 'close', 'volume']
)
print(melt_faang)
melt_faang.head()
```
