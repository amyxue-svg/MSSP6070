```python
student_performance = pd.read_csv('data/Student Performance Data.csv')
print(student_performance)
```

```python
male = student_performance[student_performance['gender'] == 'male']
```

```python
male_completed = male[male['test preparation course'] == 'completed']['total score']
```

```python
male_noncompleted = male[male['test preparation course'] == 'none']['total score']
```

```python
print(stats.normaltest(male_completed))
```

```python
print(stats.normaltest(male_noncompleted))
```

```python
print(np.mean(male_completed))
print(np.mean(male_noncompleted))
```

```python
np.mean(male_completed) - np.mean(male_noncompleted)
```

```python
tval = (np.mean(male_completed) - np.mean(male_noncompleted)) / np.sqrt(
    stats.sem(male_completed)**2 + stats.sem(male_noncompleted)**2
)
print(tval)
```

```python
SE1 = stats.sem(male_completed)
SE2 = stats.sem(male_noncompleted)
```

```python
df = (SE1**2 + SE2**2)**2 / (SE1**4 / (len(male_completed) - 1) + SE2**4 / (len(male_noncompleted) - 1))
td = stats.t(df)
p = 2 * td.sf(abs(tval))
print(p)
```
