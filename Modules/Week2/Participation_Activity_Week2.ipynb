```python
#Q5
spam = ['apples', 'bananas', 'tofu', 'cats']
def comma_code(items):
  for i in range(len(items)):
    if i == len(items)-1:
      print('and'+items[i])
    else:
      print(items[i],end=',')
comma_code(spam)
```
