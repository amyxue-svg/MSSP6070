## Exercise 4: Generate the data by running this cell
This will give you a list of numbers to work with in the remaining exercises.

```python
import random
random.seed(0)
salaries = [round(random.random()*1000000, -3) for _ in range(100)]
print(list(salaries))
```

## Exercise 5: Calculating statistics and verifying

Use the data generated above to calulate in code the following statistics without importing the statistics module. Then use the statistics module to verify your results.  Import the statistics module from https://docs.python.org/3/library/statistics.html

### mean

```python
from statistics import mean
mean(salaries)
```

### median

```python
from statistics import median
median(salaries)
```

### mode

```python
from statistics import mode
mode(salaries)
```

### sample variance
Remember to use Bessel's correction.

```python
from statistics import variance
variance(salaries)
```

### sample standard deviation
Remember to use Bessel's correction.

```python
from statistics import stdev
stdev(salaries)
```

## Exercise 6: Calculating more statistics
### range

```python
range = max(salaries) - min(salaries)
print(range)
```

### coefficient of variation

```python
coefficient_of_variation = stdev(salaries) / mean(salaries)
print(coefficient_of_variation)
```

### interquartile range

```python
from statistics import quantiles
q1,q2,q3 = quantiles(salaries,n=4)
interquartile = q3 - q1
print(interquartile)
```

### quartile coefficent of dispersion

```python
QCD = interquartile/(q1 + q3)
print(QCD)
```

## Exercise 7: Scaling data
### min-max scaling

```python

x_scale = [(x-min(salaries))/(max(salaries)-min(salaries)) for x in salaries]
print(x_scale)
```

### standardizing

```python
z_score = [((x-min(salaries))/stdev(salaries))for x in salaries]
print(z_score)
```

## Exercise 8: Calculating covariance and correlation
### covariance

```python
#since I do not see any other variables uncorrelated or correlated with x, I believe this question asking us to calculate the cov(x,x), and cov(salaries,salaries)=var(salaries)
covariance = variance(salaries)
print(covariance)
```

### Pearson correlation coefficient ($\rho$)

```python
p_x = covariance/stdev(salaries)
print(p_x)
```
